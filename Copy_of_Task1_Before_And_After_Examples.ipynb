#SMS Dataset Exploration

```python
import pandas as pd

df = pd.read_csv('SMSSpamCollection',
sep='\t', header=None, names=['label', 'message'])
```

```python
df_clean = df.drop_duplicates().copy()
```

```python
df_clean.shape
```

```python
df_clean['no_punctuation'] = df_clean['message'].str.replace(r'[^\w\s]', '', regex=True)
df_clean[['message', 'no_punctuation']].head()
```

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

df_clean['no_stopwords'] = df_clean['message'].apply(lambda text: ' '.join(word for word in text.split() if word.lower() not in stop_words))
df_clean[['message', 'no_stopwords']].head()
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

df_clean['stemmed'] = df_clean['message'].apply(lambda text: ' '.join(stemmer.stem(word) for word in text.split()))
df_clean[['message', 'stemmed']].head()
```

```python
df_clean[['message', 'no_punctuation']].head()
```

```python
df_clean[['message', 'no_stopwords']].head()
```

```python
df_clean[['message', 'stemmed']].head()
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
word_tokenize("Hello, how are you?")
```

```python
df_clean['tokens'] = df_clean['message'].apply(word_tokenize)
df_clean[['message', 'tokens']].head()
```

```python
df_clean['lowercase'] = df_clean['message'].str.lower()
df_clean[['message', 'lowercase']].head()
```

```python
df_clean['regex_cleaned'] = df_clean['lowercase'].str.replace(r'https?://\S+|www\.\S+', '', regex=True)
df_clean[['lowercase', 'regex_cleaned']].head()
```

```python
changed = df_clean[df_clean['lowercase']!= df_clean['regex_cleaned']]
print("Message changed:", len(changed))
changed[['lowercase', 'regex_cleaned']].head()
```

```python
print("Before")
print(changed['lowercase'].iloc[0])

print("\nAfter:")
print(changed['regex_cleaned'].iloc[0])
```
