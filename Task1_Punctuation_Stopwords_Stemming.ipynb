#SMS Dataset Exploration

```python
import pandas as pd

df = pd.read_csv('SMSSpamCollection',
sep='\t', header=None, names=['label', 'message'])
```

```python
df_clean = df.drop_duplicates().copy()
```

```python
df_clean.shape
```

```python
df_clean['no_punctuation'] = df_clean['message'].str.replace(r'[^\w\s]', '', regex=True)
df_clean[['message', 'no_punctuation']].head()
```

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

df_clean['no_stopwords'] = df_clean['message'].apply(lambda text: ' '.join(word for word in text.split() if word.lower() not in stop_words))
df_clean[['message', 'no_stopwords']].head()
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

df_clean['stemmed'] = df_clean['message'].apply(lambda text: ' '.join(stemmer.stem(word) for word in text.split()))
df_clean[['message', 'stemmed']].head()
```
