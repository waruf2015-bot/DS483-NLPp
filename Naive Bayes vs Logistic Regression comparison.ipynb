```python
import pandas as pd

comparison = pd.DataFrame({
"Model": ["Naive Bayes", "Logistic Regression"],
"Accuracy": [0.9671, 0.9565],
"Spam Precision": [1.0000, 0.9778],
"Spam Recall": [0.7405, 0.6718],
"Spam F1-score": [0.8509, 0.7964]
})

display(comparison)
```

```python
import matplotlib.pyplot as plt

models = comparison["Model"]

metrics = ["Accuracy", "Spam Precision", "Spam Recall", "Spam F1-score"]

comparison.set_index("Model")[metrics].plot(
kind="bar",
figsize=(10, 6)
)

plt.title("Naive Bayes vs Logistic Regression")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1.1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
best_accuracy = comparison.loc[comparison["Accuracy"].idxmax(), "Model"]
best_f1 = comparison.loc[comparison["Spam F1-score"].idxmax(), "Model"]

print("Best model based on Accuracy:", best_accuracy)
print("Best model based on Spam F1-score:", best_f1)

print("\nFinal Comparison:")
print("Naive Bayes achieved better results than Logistic Regression")
print("across all four evaluation metrics.")
```
