#SMS Dataset Exploration

```python
import pandas as pd

df = pd.read_csv('SMSSpamCollection',
sep='\t', header=None, names=['label', 'message'])
```

```python
df_clean = df.drop_duplicates().copy()
```

```python
df_clean.shape
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
word_tokenize("Hello, how are you?")
```

```python
df_clean['tokens'] = df_clean['message'].apply(word_tokenize)
df_clean[['message', 'tokens']].head()
```

```python
df_clean['lowercase'] = df_clean['message'].str.lower()
df_clean[['message', 'lowercase']].head()
```

```python
df_clean['regex_cleaned'] = df_clean['lowercase'].str.replace(r'https?://\S+|www\.\S+', '', regex=True)
df_clean[['lowercase', 'regex_cleaned']].head()
```

```python
changed = df_clean[df_clean['lowercase']!= df_clean['regex_cleaned']]
print("Message changed:", len(changed))
changed[['lowercase', 'regex_cleaned']].head()
```
