```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


df = pd.read_csv("SMSSpamCollection_cleaned.csv")


tfidf = TfidfVectorizer(stop_words="english")
X_tfidf = tfidf.fit_transform(df["message"])

print("Dataset and TF-IDF loaded successfully!")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(X_tfidf[:10], X_tfidf)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python

for i in range(10):

    original_message = df["message"].iloc[i]

    print("\nOriginal message:")

    print(original_message)

    seen = set()

    count = 0

    for idx in similarity_matrix[i].argsort()[::-1]:

        similar_message = df["message"].iloc[idx]

     

        if similar_message == original_message:

            continue

        if similar_message in seen:

            continue

        seen.add(similar_message)

        print("Similarity:", round(similarity_matrix[i][idx], 3))

        print("Similar message:", similar_message)

        count += 1

        if count == 3:

            break
```
